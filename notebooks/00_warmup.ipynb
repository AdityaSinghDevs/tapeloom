```python
class Vec2:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self): #printing
        return f"Vec2({self.x}, {self.y})"

    def __add__(self, other):
        return Vec2(self.x + other.x, self.y + other.y)

    def __mul__(self, other):
        return Vec2(self.x * other.x, self.y * other.y)
```

```python
print(Vec2(1,2) + Vec2(3,4))
print(Vec2(1,2) * Vec2(3,4))
```

```python
def make_adder(n):
    def add(x):
        return x + n
    return add
```

```python
add5 = make_adder(5)
print(add5(3))
print(make_adder(10)(3))
```

```python
# Gradient descent 
w, lr  = 0.0, 0.1

# Loss is L(w) = (w-3)^2, want lowest value which we find at w=3, 
# slope is dL/dw = 2(w-3), at w = 0 slope is -6, w increases slope goes down, so moving opposite to slope

for _ in range(200):
    grad = 2 * (w-3)
    w = w - (lr * grad)
print(w)
```

```python
# SINGLE NEURON NEURAL NETWORK

def nn(x1, x2, target, lr):

  w1, w2 = 1.0, 1.0
  for _ in range(30):
       
       y = x1*w1 + x2*w2
   
       loss = (1/2)*(target - y)**2
   
       dl_dyhat = y - target 
       dy_dw1 = x1 
       dl_dw1 = dl_dyhat * dy_dw1
   
       dl_dyhat = y - target 
       dy_dw2 = x2 
       dl_dw2 = dl_dyhat * dy_dw2
   
       #gradient descent
   
       w1_new = w1 - lr*(dl_dw1)
       w2_new = w2 - lr*(dl_dw2)
   
       w1 = w1_new
       w2 = w2_new
   
       print(f"New weights are : {w1, w2}, loss : {loss}, predicted value was : {y}")

nn(1.0,2.0,5.0,0.1)


```

```python
# SINGLE NEURON NEURAL NETWORK (BETTER IMPLEMENTATION)

def nn(x1, x2, target, lr):

  w1, w2 = 1.0, 1.0
  for _ in range(30):
       
       y = x1*w1 + x2*w2
   
       loss = 0.5*(target - y)**2
   
       dl_dyhat = y - target 


       dl_dw1 = dl_dyhat * x1
       dl_dw2 = dl_dyhat * x2
   
       #gradient descent
   
       w1 -= lr*(dl_dw1)
       w2 -= lr*(dl_dw2)
   
       print(f"Weights: {w1:.3f}, {w2:.3f}, Loss: {loss:.3f}, Prediction: {y:.3f}")

nn(1.0,2.0,5.0,0.1)


```
